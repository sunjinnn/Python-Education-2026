```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib

data = {
    '영화' : ['명량', '극한직업', '신과함께-죄와 벌', '국제시장', '괴물', '도둑들', '7번방의 선물', '암살'],
    '개봉 연도' : [2014, 2019, 2017, 2014, 2006, 2012, 2013, 2015],
    '관객 수' : [1761, 1626, 1441, 1426, 1301, 1298, 1281, 1270], # (단위 : 만 명)
    '평점' : [8.88, 9.20, 8.73, 9.16, 8.62, 7.64, 8.83, 9.10]
}
df = pd.DataFrame(data)
df
```

```python
matplotlib.rcParams['font.family'] = 'NanumGothic' # 폰트 설정 변경
matplotlib.rcParams['font.size'] = 15 # 글꼴 크기 설정
matplotlib.rcParams['axes.unicode_minus'] = False # 한글 폰트 사용 시 마이너스 부호 깨짐 방지
```

```python
#퀴즈 1: 영화데이터로 막대그래프
plt.bar(df['영화'], df['평점'])
plt.show()
```

```python
# 퀴즈 2

plt.bar(df['영화'], df['평점'])
plt.title('국내 Top 8 영화 평점 정보')
plt.xlabel('영화')
plt.xticks(rotation=90)
plt.ylabel('평점')
plt.show()
```

```python
import numpy as np
#퀴즈 3: 연도별 평균
df_group = df.groupby('개봉 연도').mean(numeric_only=True)
df_group

plt.plot(df_group.index, df_group['평점'])
plt.show()
```
