```python
#퀴즈 26번:
#리조트에서 4인 이상의 손님에게는 추가비용이 발생하는 if-else문 퀴즈 풀이

total = 2 # 총 인원
if total <= 4:
  print("추가비용 없음")
else:
  print("1인당 만원")
print("감사합니다")
```

```python
#퀴즈 27번:
#열이 39도 이상이면 고열 38도 이상 39도 미만이면 미열 그외에는 정상을 출력 하는 퀴즈 풀이

temp = 40 # 체온
if temp >= 39:
  print("고열입니다")
elif temp >= 38:
  print("미열입니다")
else:
  print("정상입니다")
```

```python
#퀴즈 27번 입력문 추가

temp=int(input("열은 몇도인가요? :"))
if temp >= 39:
  print("고열입니다")
elif temp >= 38:
  print("미열입니다")
else:
  print("정상입니다")
```
