```python
# 퀴즈:
# OpenCV 를 이용하여 가로로 촬영된 동영상을 세로로 회전하는 프로그램을 작성하시오

import cv2  # OpenCV 라이브러리 불러오기

# city.mp4 동영상 파일 열기
cap = cv2.VideoCapture('city.mp4')

# 저장할 동영상의 코덱을 DIVX로 설정
fourcc = cv2.VideoWriter_fourcc(*'DIVX')

# 원본 동영상의 가로 길이 가져오기
width = round(cap.get(cv2.CAP_PROP_FRAME_WIDTH))

# 원본 동영상의 세로 길이 가져오기
height = round(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# 원본 동영상의 FPS(초당 프레임 수) 가져오기
fps = cap.get(cv2.CAP_PROP_FPS)

# 새로운 동영상 파일을 저장하기 위한 객체 생성
# fps * 4 : 원본보다 4배 빠르게 재생
# 90도 회전하면 가로와 세로가 바뀌므로 (height, width)로 설정
out = cv2.VideoWriter(
    'city_output.avi',
    fourcc,
    fps * 4,
    (height, width)
)

# 동영상 파일이 정상적으로 열려 있는 동안 반복
while cap.isOpened():

    # 동영상에서 한 프레임씩 읽어오기
    # ret : 프레임을 정상적으로 읽었는지 여부
    # frame : 실제 이미지 데이터
    ret, frame = cap.read()

    # 더 이상 읽을 프레임이 없으면 반복문 종료
    if not ret:
        break

    # 현재 프레임을 시계 반대 방향으로 90도 회전
    rotate_frame = cv2.rotate(
        frame,
        cv2.ROTATE_90_COUNTERCLOCKWISE
    )

    # 회전된 프레임을 새로운 동영상 파일에 저장
    out.write(rotate_frame)

    # 원본 동영상을 화면에 출력
    cv2.imshow('video', frame)

    # q 키를 누르면 동영상 재생 중지
    if cv2.waitKey(1) == ord('q'):
        break

# 저장 중이던 동영상 파일 닫기
out.release()

# 불러온 원본 동영상 파일 닫기
cap.release()

# OpenCV로 열었던 모든 창 닫기
cv2.destroyAllWindows()
```
